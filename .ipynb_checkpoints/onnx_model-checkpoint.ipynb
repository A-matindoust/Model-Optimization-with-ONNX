```python
import tf2onnx
import tensorflow as tf
```

```python
model = tf.keras.models.load_model('mnist.keras')
```

```python
spec = [tf.TensorSpec((None,28,28,1),tf.float32)]
model_onnx,_ = tf2onnx.convert.from_keras(model,input_signature=spec,
                                          output_path='mnist.onnx')
```

```python
model_onnx.graph.output[0].name
```
